```python
trojkat_pascala = []

n = int(input('Podaj n'))

for i in range(1, n+1):
    liczby = []
    for j in range(0, i):
        if j == 0 or j == i-1:
            liczby.append(1)
        else:
            liczby.append(trojkat_pascala[i-2][j-1] + trojkat_pascala[i-2][j])
    trojkat_pascala.append(liczby)

szerokosc = 4

for linia in trojkat_pascala:
    padding = (' ') * (n - len(linia))
    print(padding + ' '.join([str(n) for n in linia]))
   
```

```python
liczby = [input()]
liczby = sorted(liczby)
liczbaprzesuniec = 1
for n in liczby:
    if n == ' ':
        del liczby[liczbaprzesuniec]
        liczbaprzesuniec += 1

liczby_pierwsze = []
for i in  liczby:
    def liczba_pierwsza(liczba):
        if liczba != 1 and liczba != i:
            if liczba % i != 0:
                return liczba
    if liczba_pierwsza(i) == True:
        liczby_pierwsze.append(i)
        
liczby_pierwsze
```

```python
def czy_liczba_pierwsza(liczba):
    for i in range(2, liczba -1):
        if liczba % i == 0:
            return False
    return True
```

```python
czy_liczba_pierwsza(8)
```

```python
liczby = input()
liczby_pierwsze = []
liczby = [int(n) for n in liczby.split()]

for liczba in liczby:
    if czy_liczba_pierwsza(liczba) == True:
        liczby_pierwsze.append(liczba)
liczby_pierwsze
```

```python
liczba = int(input('podaj liczbę'))
dzielniki = []
def liczba_doskonala(liczba):
    for i in range(1, liczba):
        if liczba % i == 0:
            dzielniki.append(i)
    if sum(dzielniki) == liczba:
        print('ta liczba jest doskonala')
    else:
        print('ta liczba nie jest doskonala')
        
            
```
